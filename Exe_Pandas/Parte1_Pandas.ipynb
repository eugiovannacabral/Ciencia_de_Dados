```python
labels1 = ['Java', 'C', 'Python']
valores1 = [16.25, 16.04, 9.85]

labels2 = ['C', 'Python', 'Java']
valores2 = [16.21, 12.12, 11.68]

seriesAno1 = pd.Series(index=labels1, data=valores1)
seriesAno2 = pd.Series(index=labels2, data=valores2)

soma_valores1 = seriesAno1.sum()
soma_valores2 = seriesAno2.sum()

print("Porcentagem da seriesAno1:", round(soma_valores1, 2), "%")
print("Porcentagem da seriesAno2:", round(soma_valores2, 2), "%")
```

```python
diferenca = seriesAno2-seriesAno1
print("Porcentagem de declínio/crescimento de cada linguagem", diferenca)
```

```python
cond = diferenca > 0
print("Linguagens que tiveram crescimento: ")
print(diferenca[cond])
```

```python
diferenca2 = seriesAno2 + diferenca * 2 #vai ser o valor do quarto ano
print(diferenca2.nlargest(1))
```

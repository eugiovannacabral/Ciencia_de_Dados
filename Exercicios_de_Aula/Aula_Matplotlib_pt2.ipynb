#Fundamentos de Seaborn

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#importando o dataset tips
ds_tips = sns.load_dataset('tips')
print(ds_tips)

#setando um estilo diferente no gráfico
sns.set_style('dark')
#setando uma fonte maior
sns.set_context('talk')

#tracando um scatterplot, scatter quer dizer dispersão
sns.scatterplot(data = ds_tips, x= 'total_bill', y ='tip')
#precisa da ajuda do matplotlib pois o seaborn mais simples e objetivo
plt.xlabel('Conta total em US$')
plt.ylabel('Gorjeta em US$')
plt.show()
```

#Hsistograma

```python
#importando o dataset penguins
ds_penguins = sns.load_dataset('penguins')
print(ds_penguins.columns)

#traçando o histograma
sns.histplot(
    data= ds_penguins, #passando o dataset
    x = 'flipper_length_mm', # o histograma já divide os dados automaticamente em bins/LATINHAS/, já cria os intervalos
    hue = 'species', # o hue só usa dados categóricos
    kde = 'True'
)
```

#Boxplot

```python
ds_tips = sns.load_dataset('tips')
#trancado um boxplot
sns.boxplot(
    data = ds_tips,
    x = 'day',
    y = 'tip',
    hue = 'sex'
)
plt.show()
```

#Regplot

```python
sns.regplot(
    data = ds_tips,
    x = 'total_bill',
    y = 'tip',
    line_kws = {'color':'red'} #essa linha traçada no meio ajuda a verificar a tendência dos dados
)
plt.show()
```

#Heatmap

```python
#selecionando as colunas que quero identificar uma correlação
corr = ds_tips[['total_bill', 'tip', 'size']].corr()

#tracando o heatmap
sns.heatmap(
    corr,
    annot = True,
    fmt = '.2f'
)
plt.show()
```

##Slicing no NumPy

```python
import numpy as np
```

```python
#criando a matriz
#palntando a mesma semente aleatória
np.random.seed(10)

#pedindo para criar uma matriz com nove números aleatórios dentro desse intervalo, depois restruturamos a matriz
mtz =  np.random.randint(1,99,9).reshape(3,3)
print(mtz)

```

```python
#Extraindo apenas a segunda linha da matriz
print(mtz[1])
#Extraindo apenas a terceira coluna da matriz, os : pega todos os elementos
print(mtz[:, 2])
#Extraindo a matriz 2x2 no canto inferior direito da amtriz original
print(mtz[1:,1:])#está dizendo quero os elementos da fileira 1 e coluna 1 para a frente
```

##Condicionais no Numpy

##Análise de padrões textuais com NumPy

```python
arr = np.array(["Inatel", 'CasaViva', 'ICC', 'CDG', 'eHealth', 'CSILab', 'RobotBulls', 'ProdLab', 'CRA', 'CRR'])
print(arr)

#padroniza todo mundo para maiusculo para não ter problema na hora de verificar se é minusculo ou maisculo
arr = np.char.upper(arr)

#submódulo do NumPy para tarbalhar com texto: char
#buscando qual texto aceita um padrão informado
print(np.char.find(arr, 'A')) # vai mostrar números indicando a posição da primeira vez que aparece 'a', vai mostrar -1 quando não atender o padrão
print(np.char.find(arr, 'A')>=0)

cond = np.char.find(arr, 'A')>=0
print(arr[cond])
```

##Como importar Datasets no Numpy

```python
#carregando o dataset e avisando que o delimitador que separa um dado do outro dentro do arquivo é o ';'
dataset = np.loadtxt('space.csv', delimiter = ';', dtype = 'str')
print(dataset)
```

```python
#extraindo as colunas do dataset
print(dataset[0,:]) #as colunas estão no top do dataset por isso colocou para mostrar a linha 0

#extraindo o nome das empresas
print(dataset[1:, 1])
print(np.unique(dataset[1:, 1], return_counts = True)) #extraindo e contando os elementos únicos
```

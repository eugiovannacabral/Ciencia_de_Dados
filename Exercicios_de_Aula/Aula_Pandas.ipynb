#PANDAS SERIES(1-D)

```python
import numpy as np
import pandas as pd
```

```python
#Preencher uma series
#lista de labels
#lista de valores

labels = ['Tiago', 'Mateus','Bruna','Julia'] #etiqueta
valores = [23,25,27,22]

#criando a series
se1 = pd.Series(index=labels, data=valores) # uma series parece uma associação, podemos passar direto uma estrutura de dicionário
print(se1)
print(type(se1))

#acessando elementos da series
print(se1['Mateus'])
print(se1[['Mateus', 'Julia']])
```

#Operações entre series


```python
s1 = pd.Series({'v1':10, 'v2':45, 'v3':70})
s2 = pd.Series({'v1':10, '42':50, 'v3':70})

#print(s1,s2)
print(s1+s2)

print(s1.add(s2, fill_value = 0))#pode costumizar o valor que vai utilizar, porque quando não existe umm par correspondendo ele simplemente não faz a operação
```

#Pandas dataframe(2d)

```python
#Como preencher um dataframe, que é como uma planilha de excell que não tem indez só nas colunas, mas nas linhas também
#lista de labels(colunas)
colunas = ['w','x','y','z']
#lista de labels(linhas)
linhas = ['a','b','c','d','e']
#lista de valores
np.random.seed(10)
valores = np.random.randint(1,50,[5,4])

np.random.randint(1,50,[5,4])

df = pd.DataFrame(columns = colunas, index = linhas,data = valores)
print(df)
```

#Manipulando o dataframe

```python
print(df)

#puxando uma única coluna do dataframe
print(df['x'])

#puxando 2 colunas do dataframe
print(df[['y','z']]) #para ele poder pegar 2 valores precisa passar uma lista para ele

#puxando uma única célula
print(df['y']['c'])

#puxando múltiplas colunas
print(df[['w','x','z']])
```

#Slicing de dados no dataframe com loc(labels) e iloc (indices numericos)

```python
print(df)

#puxando uma única linha do dataframe
print(df.loc['c',['w','x','y','z']])
#ou
print(df.iloc[2, :])

#puxando multiplas linhas
print(df.loc[['b','e'], ['w','x','y','z']])
#ou
print(df.iloc[[1,4], :])


#puxando a matriz 2x2 do canto inferior direito
print(df.loc[['d','e'], ['y','z']]) #loc vem de location
print(df.iloc[[3,4], [2,3]])
```

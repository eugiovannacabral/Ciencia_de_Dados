# Fundamentos de Matplotlib


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Plotando gráficos de linhas com MatplotLib (Plot, ou ainda Lineplot)

```python
# eixo x
x = np.array([1,2,3,4])
# eixo y
y = x*2

# eixo y2
y2 = x**2

# legendas nos eixos x e y (comentamos para poder utilizar a legenda para cada subplot)
#plt.xlabel('Valores X')
#plt.ylabel('Valores Y')

# Quando se cria um subplot deve-se passar QUANTAS LINHAS e QUANTAS COLUNAS tem a grade subplots
plt.subplot(1, 2, 1)
plt.plot(x, y, '*:r', linewidth = '3', markersize = 20)
plt.xlabel('Valores X')
plt.ylabel('Valores Y')

plt.subplot(1, 2, 2)
plt.plot(x, y2, 's--b', linewidth = '3', markersize = 20)
plt.xlabel('Valores X')
plt.ylabel('Valores Y')
plt.show()
```

Scatter Plot (Gráfico de Dispersão)

```python
ds = pd.read_csv('paises.csv', delimiter=';')
#print(ds.columns)

# Quão dispersas estão as rendas per capita dos 6 maiores países do mundo?
# Pegando apenas os 6 maiores países do planeta
maioresPaises = ds.nlargest(6, 'Area (sq. mi.)')
#print(maioresPaises)

# Traçando o gráfico com as rendas per capita destes países
# O parâmetro 's' permite adicionar uma terceira dimensão neste gráfico
plt.scatter(maioresPaises['Country'],
            maioresPaises['GDP ($ per capita)'],
            s = maioresPaises['Area (sq. mi.)']/10000)
plt.show()
```

Bar Plot (Gráfico de Barras)

```python
# Qual a diferença das 5 maiores rendas per capita do dataset?
maioresGDP = ds.nlargest(5, 'GDP ($ per capita)')
print(maioresGDP)

plt.bar(maioresGDP['Country'], maioresGDP['GDP ($ per capita)'])
plt.show()
```

Pie Plot (Gráfico de Pizza)

```python
# Pegando os países que não possuem costa marítima
paisesSemCosta = ds[ds['Coastline (coast/area ratio)'] == 0]
print(paisesSemCosta)

# Calculando a quantidade de países que não possuem costa e a quantidade de países que possuem costa
qtSemCosta = len(paisesSemCosta)
qtComCosta = len(ds) - qtSemCosta

# traçando o gráfico em torta/pizza
plt.pie(x = [qtSemCosta, qtComCosta], labels=['% Países sem Costa',
                                              '% Países com Costa'],
        autopct='%1.1f%%')
plt.show()
```

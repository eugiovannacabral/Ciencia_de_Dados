```python
import numpy as np
import pandas as pd
```

```python
ds = pd.read_csv('paises (1).csv',sep = ';')
#print(ds)
print(ds.columns) # visualizando as colunas
print(ds.head(5)) # extraindo registros do topo
print(ds.tail(3)) #extraindoregistros da base
```

#Slicing no dataset pandas

```python
print(ds.columns)
print(ds[['Region', 'Climate','Service',]])
```

#Criando novos arquivos com pandas

```python
#calculando a porcentagem da populacao de cada pais do mundo
total_population = np.sum(ds['Population']) #calculando a população total do planeta
print(total_population)

#calculando a porcentagem de cada pais
seriesPorcPaises =(ds['Population'] / total_population)* 100 #broadcasting
print(seriesPorcPaises)

#adicionar esta Series no Dataset
ds['% Population'] = np.round(seriesPorcPaises,3)

#criando uma v2 do dataset com uma coluna nova, foi como se criasse uma copia do dataset e adicionasse uma nova coluna
ds.to_csv('paises_v2.csv') #isso é uma boa prática

#pegando os 5 paises que mais tem gente
print(ds.nlargest(5, '% Population') ['Country'])

#pegando os 5 menores paises
print(ds.nsmallest(5, '% Population') ['Country'])
```

#Agrupamento de dados no Pandas

obs: só funciona se o dataset tiver ao menos uma coluna de dados categóricos

```python
#agrupando por região
#contar o numero de paises por regiao
group_region = ds.groupby('Region')
print(group_region.count() ['Country'])

#somando a populacao de cada regiao
group_region = ds.groupby('Region')
print(group_region.sum()['Population'])
```

#Criação e aplicação de funções customizadas com o pandas

```python
#Criando uma função no Python
#criar função para dar 10% de desconto em alguma coisa
def tenPercent (x):
  return x * 0.9
```

```python
#Buscando a coluna que guarda a taxa de natalidade de um pais
print(ds.columns)

taxa_mortalidade = ds['Deathrate']
print(taxa_mortalidade)

ds['Deathrate -10%'] = taxa_mortalidade.apply(tenPercent)
print(ds.head(2))
```

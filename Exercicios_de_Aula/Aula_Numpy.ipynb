# Criando um Numpy Array

```python
import numpy as np # esse np é um apelido ao numpy
```

```python
#criando um numpy array 1 dimensão
arr = np.array([10,20,30,40,50,60])#a lista só alimentou o numpy array
print(arr)
print(type(arr))
print(arr.size)#tamanho do array
print(arr.ndim)#dimensão do array
print(arr.shape)#tamanho

mtz = np.array([[10,20],[30,40],[50,60]])#matriz
print(mtz)
print(mtz.size)
print(mtz.ndim)
print(mtz.shape)
```

# Funções pré-prontas para estruturarmos numpy arrays

```python
#Ones
mtz = np.ones([5,5])
print(mtz)

#Zeros
arr = np.zeros(10)
print(arr)

#transfromando uma array uni dimension e two dimension(matriz)
print(arr.reshape(5,2))

#Arange(alcance do array)
mtz = np.arange(2, 21, 2)
print(mtz)
print(mtz.reshape(2, 5))
```

# Operações entre numpy arrays

```python
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([60,40, 20, 10, 5])
arr3 = arr1 + arr2

#Operações entre arrays
#somas de elementos de cada array de acordo com cada indice, no java a gente fazia um for e comparava se o indice do arr1 for igual ao do arr2 soma seus elementos
print(arr3)
print(arr1 - arr2)
print(arr1 * arr2)

#Concatenação de arrays
arr3 = np.concatenate([arr1, arr2])
print(arr3)

#Broadcasting - quando um escalar faz uma operação com um array
print(5 * arr3)
```

```python
#estruturando uma matriz com contas
mtz = np.arange(10,96, 5)
mtz = mtz.reshape(3, 6)
print(mtz)

#Extraindo a soma da primeira coluna
print(mtz.sum(axis = 0)) #Eixo 0 = coluna, soma de todas as colunas
print(mtz.sum(axis = 0) [0]) #só da primeira coluna

#Extraindo a soma da segunda linha
print(mtz.sum(axis = 1)[1])
```

# Números aleatórios no numpy(módulo random)

```python
#Platando a mesma semente aleatória, quando quer manipular o algoritmos de números aleatóris a partir do mesmo princípio que eles foram criados
np.random.seed(5)

arr = np.random.randint(1, 10, 10) # escolha 10  números aleatórios dentro de 1 a 10
print(arr)

#extraindo elementos únicos de um vetor, ordena os elementos
print(np.unique(arr))

#Contando elementos únicos
print(np.unique(arr, return_counts = True)) #retornou duas tuplas
```

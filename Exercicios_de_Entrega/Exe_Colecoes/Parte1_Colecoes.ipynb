```python
times = ['Barcelona', 'Flamengo', 'Cruzeiro', 'Bahia', 'Vasco']

print(times[0:3])
print(times[3:])

times_ordenados = sorted(times) # função organiza os itens e retorna uma nova lista sem alterar e 'o sort ' é o método que reogarniza a lista em ordem alfabética
print(times_ordenados)

posicao = times.index('Barcelona') #para pegar o índice do elemento dentro da lista
print(posicao)
```

```python
loja1 = {'Iphone12', 'Iphone13', 'Iphone14', 'Iphone15'}
loja2 = {'SamsungA34', 'SamsungA57', 'SamsungS11', 'SamsungS12', 'Iphone12'}
print(loja1, loja2)

lojas = loja1 | loja2 #união entre conjuntos
print(lojas)

ambas_lojas = loja1 & loja2 #interseção entre conjuntos
print(ambas_lojas)

```

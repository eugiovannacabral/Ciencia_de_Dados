```python
nomes = []
pesos = []

for i in range(3):
  nome = input("Qual seu nome? ")
  peso = int(input("Qual o seu peso?"))

  nomes.append(nome)
  pesos.append(peso)

pesado = max(pesos)
leve = min(pesos)
posicao_leve = pesos.index(leve) #pega o index da lista
posicao_pesado = pesos.index(pesado)

print("A pessoa mais leve é", nomes[posicao_leve])
print("A pessoa mais pesada é", nomes[posicao_pesado])
```

```python
n = int(input("Sobre quantas pessoas deseja registrar? "))
nomes = []
idades = []
sexos = []

for i in range(n):
  nome = input("Qual seu nome? ")
  idade = int(input("Qual a sua idade?"))
  sexo = input("Qual seu sexo? ")

  nomes.append(nome)
  sexos.append(sexo)
  idades.append(idade)

soma = sum(idades)
media = soma/n
print("Média de suas idades: ", round(media, 2))

quant = 0
for i in range(n):
    if idades[i] < 20 and sexos[i] == 'F':
        quant += 1

print("Quantidade de mulheres com menos de 20 anos:", quant)
```

```python
mercado = []
estoque = []

for i in range(3):
  nome = input("Qual o nome do seu produto? ")
  preco = float(input("Qual é o preço? "))
  quant = int(input("Qual é a quantidade em estoque? "))

  produtos = {
    "Nome": nome,
    "Preco": preco,
    "Quantidade em estoque": quant
  }

  valor_estoque = preco * quant

  mercado.append(produtos)
  estoque.append(valor_estoque)

for i in range(3):
  print(mercado[i]['Nome'], "estoque:",  estoque[i])
```

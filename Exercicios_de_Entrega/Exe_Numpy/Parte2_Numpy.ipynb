```python
import numpy as np
```

```python
mtz = np.array([[5,2],[7,4],[9,15],[32,67],[1,9]])
print(mtz)
print(mtz.shape)#tamanho

tamanho = mtz.shape
multiplicacao = tamanho[0] * tamanho[1]

if(multiplicacao % 2 == 0):
  print("Pode se tornar unidimensional com um número par de elementos")
else:
  print("Pode se tornar unidimensional com um número ímpar de elementos")
```

```python
np.random.seed(10) # para padronizar os números aleatórios
mtz = np.random.randint(1, 51,[4, 4])#matriz 4x4 formada por números de 0 até 50
print(mtz)

media_linhas = np.mean(mtz, axis=1)
media_colunas = np.mean(mtz, axis=0)
print("Média das linhas:", media_linhas)
print("Média das colunas:", media_colunas)

print(np.unique(mtz)) #mostra o números únicos do array

#mostra quantas vezes estes apareceram
numeros, quantidades = np.unique(mtz, return_counts=True)

for i in range(len(numeros)):
    if quantidades[i] == 2:
        print('Números que aparecem 2 vezes:',numeros[i])
```

```python
arr1 = np.array([1,1,1,1,1,1,1,1])
arr2 = np.random.randint(1, 9, 8)
arr3 = arr1 + arr2

soma_total = arr3.sum()
if(soma_total >= 40):
  print(arr3.reshape(4,2))
else:
  print(arr3.reshape(2,4))
```

```python
arr1 = np.arange(0, 51, 2)
arr2 = np.arange(100,50, -2)

arr3 = np.concatenate([arr1, arr2])
arr3_ordenado = np.sort(arr3) #método que reordena o própio array
print(arr3_ordenado)
```

```python
import numpy as np
```

```python
dataset = np.loadtxt('space.csv', delimiter = ';', dtype = 'str')
#print(dataset)
#print(dataset[0,:])colunas

status = dataset[1:, 7] #pega todas as informações da coluna 7

sucessos = np.sum(status == 'Success') # o np.sum() conta os True como 1.
total = len(status) #pega o número total de missões.

porcentagem = (sucessos / total) * 100

print(porcentagem,"% missões deram certo!")
```

```python
custos = dataset[1:, 6]
print(custos)

custos = custos.astype(float)#pq já haviamos definido o dataset para ser tudo string
maiores = custos[custos > 0]

print("Média das missões maiores que",round(np.mean(maiores), 2))
```

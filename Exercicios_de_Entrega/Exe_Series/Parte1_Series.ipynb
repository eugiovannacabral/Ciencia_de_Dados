```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
d1 = pd.read_csv('airtravel.csv', delimiter=",",
                 index_col='Date', parse_dates=True)

d2 = pd.read_csv('co2_emissions.csv', delimiter=",",
                 index_col='Year', parse_dates=True)

print(d1.columns)
print(d2.columns)

# Plotando a série temporal de passageiros
d1['Passengers'].plot(
    figsize=(8, 6),
    title='Passageiros em viagens aéreas ao longo do tempo',
    xlabel='Data',
    ylabel='Número de passageiros',
    x_compat=True
)
plt.show()

# Respostas das perguntas
# a) Sim, a série possui tendência crescente
# b) Sim, sazonalidade anual, a cada 12 meses
# c) Não tem dados o suficiente para analizar o ciclo. As oscilações do gráfico estão muito relacionadas a sazonalidade anual

# Plotando a série temporal de emissões de CO₂
d2['CO2_Emissions'].plot(
    figsize=(8, 6),
    title='Emissões de CO₂ ao longo do tempo',
    xlabel='Ano',
    ylabel='Emissões de CO₂',
    x_compat=True
)
plt.show()

# a) Sim. A tendência é decrescente, aproximadamente linear:
# as emissões diminuem ao longo dos anos.

# b) Não é possível identificar sazonalidade dentro de cada ano,
# pois os dados são anuais, sem informações mensais ou trimestrais.

# c) Há indícios de um ciclo de aproximadamente 8 anos,
# com picos próximos de 1992, 2000, 2008 e 2016.
# Isso indica oscilações de vários anos em torno da tendência.
# Apenas esses dados não permitem determinar a causa do ciclo.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ds_iris = sns.load_dataset('iris')
#print(ds_iris)
cond = ds_iris['species'] == 'setosa'
#print(ds_iris[cond])
ds_iris_setosa = ds_iris[cond]

corr = ds_iris_setosa[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr() #criando a matriz de correlação
sns.heatmap(
    corr,
    annot = True,
    fmt = '.2f'
)
plt.show()
```

```python
ds_titanic = sns.load_dataset('titanic')
#print(ds_titanic)

#traçando o histograma
sns.histplot(
    data= ds_titanic, #passando o dataset
    x = 'age', # o histograma já divide os dados automaticamente em bins/LATINHAS/, já cria os intervalos
    hue = 'sex', # o hue só usa dados categóricos
    kde = 'True',
    palette={"male": "#00008B", "female": "#FF69B4"}
)
```

```python
#trancado um boxplot
sns.boxplot(
    data = ds_titanic,
    x = 'class',
    y = 'age',
    hue = 'sex'
)
plt.show()
```

```python
ds_mpg = sns.load_dataset('mpg')
#print(ds_mpg)

sns.regplot(
    data = ds_mpg,
    x = 'mpg',
    y = 'horsepower',
    line_kws = {'color':'red'} #essa linha traçada no meio ajuda a verificar a tendência dos dados
)
plt.show()
```

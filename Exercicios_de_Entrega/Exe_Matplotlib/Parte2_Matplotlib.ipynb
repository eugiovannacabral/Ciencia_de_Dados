```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ds1 = pd.read_csv('paises (2).csv',sep = ';')
ds2 = pd.read_csv('space.csv',sep = ';')
```

```python
america_latina = ds1['Region'].str.strip() == 'LATIN AMER. & CARIB'

dados_latina = ds1[america_latina]

x = dados_latina['GDP ($ per capita)']
y = dados_latina['Literacy (%)']
populacao = dados_latina['Population']

#gráfico de dispersão
plt.scatter(
    x,
    y,
    s=populacao / 100000,
    color='purple'
)

plt.xlabel('GDP ($ per capita)')
plt.ylabel('Literacy (%)')
plt.title('GDP x Alfabetização - América Latina')

plt.show()
```

```python
status = ds2['Status Rocket'].value_counts()

plt.pie(
    status.values,
    labels=status.index,
    autopct='%1.1f%%'
)

plt.title('Status dos foguetes')

plt.show()
```

```python
europa = ds1['Region'].str.strip() == 'WESTERN EUROPE'
dados_europa = ds1[europa]

gdp = dados_europa['GDP ($ per capita)']
phones = dados_europa['Phones (per 1000)']

plt.plot(
    dados_europa['Country'],
    gdp,
    'o-r',
    linewidth=2
)

plt.plot(
    dados_europa['Country'],
    phones,
    's--b',
    linewidth=2
)

plt.xlabel('Países')
plt.ylabel('Valores')
plt.title('GDP x Phones - Europa Ocidental')

plt.legend(['GDP per capita', 'Phones per 1000'])

plt.xticks(rotation=90)

plt.show()
```

```python
sucesso = ds2[ds2['Status Mission'] == 'Success']
falha = ds2[ds2['Status Mission'] == 'Failure']

top_sucesso = sucesso['Company Name'].value_counts().head(5)
top_falha = falha['Company Name'].value_counts().head(5)


plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.bar(top_sucesso.index, top_sucesso.values)

plt.title('5 empresas com mais sucessos')
plt.ylabel('Quantidade de missões')
plt.xticks(rotation=45)


plt.subplot(1, 2, 2)

plt.bar(top_falha.index, top_falha.values)

plt.title('5 empresas com mais falhas')
plt.ylabel('Quantidade de missões')
plt.xticks(rotation=45)


plt.tight_layout()
plt.show()
```

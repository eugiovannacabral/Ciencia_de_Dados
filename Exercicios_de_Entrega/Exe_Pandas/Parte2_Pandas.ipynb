```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(10)

df = pd.DataFrame(index = ['A','B','C','D','E'], columns = ['W','X','Y','Z'], data = np.random.randint(1,50,[5,4]))
print(df)

cond1 = df['X']
print(cond1)

cond2 = cond1[cond1 < 30]
print(cond2)

print(cond2.mean())
```

```python
cond1 = df.loc['D']
print("A média dos elementso da linha D: ", cond1.mean())

cond2 = df.iloc[4]
print("A soma dos elementso da linah E: ", cond2.sum())
```

```python
filtro = df.loc[['A','C','E'], ['X','Y']]
print(filtro)
soma_linha = filtro.sum(axis=1)  #linha
print(soma_linha)

soma_coluna = filtro.sum(axis=0)  #coluna
print(soma_coluna)
```

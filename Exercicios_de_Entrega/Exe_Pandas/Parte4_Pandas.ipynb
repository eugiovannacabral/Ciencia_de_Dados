```python
import numpy as np
import pandas as pd

ds = pd.read_csv('paises (2).csv',sep = ';')
```

```python
group_region = ds.groupby('Region')
print((group_region['Population'].describe()).head())
```

```python
def reduzir_mortalidade(valor):
    return valor * 0.85

coluna = ds['Infant mortality (per 1000 births)'].apply(reduzir_mortalidade)
coluna.name = 'Infant mortality reduzida'

resultado = pd.concat([
    ds['Infant mortality (per 1000 births)'],coluna], axis=1)
print(resultado)
```

```python
ds = ds.drop('Coastline (coast/area ratio)', axis=1)

ds.to_csv('paises_sem_coastline.csv', index=False)
```
